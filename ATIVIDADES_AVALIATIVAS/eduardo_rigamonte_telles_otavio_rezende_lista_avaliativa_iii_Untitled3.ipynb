```python
#01 - Crie uma lista com 15 números inteiros digitados pelo usuário.
#Em seguida, remova todos os números repetidos, mantendo a ordem original.

numeros = []

print("Digite 15 números inteiros:")
for i in range(15):
    num = int(input(f"Digite seu {i+1}° número: "))
    numeros.append(num)

numeros_sem_repetir = []

for num in numeros:
    if num not in numeros_sem_repetir:
        numeros_sem_repetir.append(num)

print(f"\nLista Original - {numeros}")
print(f'\nLista sem números repetidos - {numeros_sem_repetir}')
```

```python
#02 - Escreva um programa que leia 10 nomes e 10 idades.
#Armazene os nomes em uma lista e as idades em outra. Em seguida, exiba o nome da pessoa mais velha.

nomes = []
idades = []
print("Digite 10 nomes:")

for i in range(10):
    nome = input(f"Digte o {i+1}° nome:  ").capitalize()
    idade = int(input(f"Digite a idade referente ao {i+1}° nome: "))
    nomes.append(nome)
    idades.append(idade)
    print("\n")
maior_idade = max(idades)
posicao_maior_idade = idades.index(maior_idade)
nome_da_pessoa = nomes[posicao_maior_idade]

print(f'A pessoa mais velha dentre as que estão na lista é {nome_da_pessoa} com a idade de {maior_idade} anos.')
```

```python
#03- Faça um programa que leia uma lista de 10 números e divida todos os valores ímpares por 3 e os pares por 2, substituindo os valores originais.
 #(Use estruturas condicionais.)

numeros = []

for i in range(10):
    num = int(input(f"Digite o {i+1}° número: "))
    numeros.append(num)
    numeros.sort()
print('\n')
print(numeros)
for i in range(len(numeros)):
    if numeros[i] % 2 == 0:
        numeros[i] = numeros[i] / 2
    else:
        numeros[i] = numeros[i] / 3
print(f'\n A lista atualizada é {numeros}.')
```

```python
#04 - Solicite ao usuário uma frase
#Armazene  em uma lista cada palavra que começa com uma vogal.
#Exiba a lista resultante.

termo = []

frase = input("Digite uma frase de sua preferencia: ").capitalize()
palavras = ""
vogais = 'aeiouAEIOUÁÉÍÓÚáéíóú'
frase += " "
for letra in frase:
    if letra != " ":
        palavras += letra
    else:
        if len(palavras) > 0 and palavras[0] in vogais:
            termo.append(palavras)
        palavras = ""

print("\n")
print(f"As palavras da frase: '{frase}', que começam com vogais estão na lista a seguir.")
print("\n")
print(termo)
```

```python
#05 - Leia 10 números inteiros e armazene-os em uma lista.
#Em seguida, crie uma segunda lista apenas com os números primos da primeira.

numeros = []
numeros_primos = []

for i in range(10):
    elemento = int(input(f"Digite seu {i+1}° núemro: "))
    numeros.append(elemento)
    numeros.sort()
for numero in numeros:
    if numero > 1:
        primo = True
        for i in range(2,numero):
            if numero % i == 0:
                primo = False
                break
        if primo:
            numeros_primos.append(numero)
            numeros_primos.sort()
print(f'\nA lista com todos os número digitados é a seguinte: {numeros}.')
print('\nNa lista a seguir, estão somente os números primos dentre todos da lista anterior.')
print(f'{numeros_primos}')
```

```python
#06 - Escreva um programa que leia uma lista de 8 elementos e verifique (estrutura condicional) se ela é palíndroma (lê-se igual de trás para frente).

todas_palavras = []
palavras_palindromas = []

for i in range(8):
    palavras = input(f'Digite a sua {i+1}ª palavra: ')
    todas_palavras.append(palavras)

    palavra_ao_contrario = ''
    for letra in palavras:
        palavra_ao_contrario = letra + palavra_ao_contrario
    if palavras == palavra_ao_contrario:
        palavras_palindromas.append(palavras)
print(f'\nDentre as palavras {todas_palavras}, por vc digitadas, ')
print(f'\n{palavras_palindromas}, essa são palíndromas.')
```

```python
#07- Crie duas listas com 10 números inteiros aleatórios entre 1 e 100.
#Gere uma terceira lista contendo apenas os elementos que estão presentes em ambas (interseção via estruturas de repetição).

numeros_aleatorios_1 = []
numeros_aleatorios_2 = []

print('Digite os 10 numros de 1 a 100 para a primeira lista.')
print("\n")
for i in range(10):
    while True:
        numero = int(input(f"Digite o {i+1}° número da 1ª lista: "))
        if 1 <= numero <= 100:
            numeros_aleatorios_1.append(numero)
            break
        else:
            print("Número inválido! O número deve está entre 1 e 100.")
print('\n')
for i in range(10):
    while True:
        numero = int(input(f"Digite o {i+1}° número da 2ª lista: "))
        if 1 <= numero <= 100:
            numeros_aleatorios_2.append(numero)
            break
        else:
            print('Número inválido! O número deve está entre 1 e 100.')

numeros_intersecao = []

for numero in numeros_aleatorios_1:
    if numero in numeros_aleatorios_2 and numero not in numeros_intersecao:
        numeros_intersecao.append(numero)

print(f"\nA sua primeira lista é {numeros_aleatorios_1}.")
print(f'\nA sua segunda lista é {numeros_aleatorios_2}.')
print(f'\nOs números comuns às duas listas são: {numeros_intersecao}')
```

```python
#08 - Faça um programa que simule o sorteio de 6 números da Mega-Sena (entre 1 e 60) sem repetição.
# Armazene em uma lista e exiba.

todos_numeros = []

for i in range(1,61):
    todos_numeros.append(i)

mega = []
posicao = 0
intervalo = 7

for i in range(6):
    posicao = (posicao + intervalo) % len(todos_numeros)
    sorteado = todos_numeros[posicao]
    mega.append(sorteado)

print(f"Os números da Mega-Sena são: {mega}")
```

```python
# 12. Gere uma matriz 5×5 com números aleatórios entre 1 e 99.
# Calcule a média dos elementos de cada linha e armazene os resultados em uma nova lista.

matriz = []

print('Quais são os valores da sua matriz 5x5?')
for i in range(5):
    linha = []
    for k in range(5):
        while True:
            valores = float(input(f"ELEMENTO - [{i}][{k}]: "))
            if 1 <= valores <= 99:
                break
            else:
                print("O número que vc acabou de digitar não está entre 1 a 99!")
        linha.append(valores)
    matriz.append(linha)
print(f"\nA sua matriz é: ")
for linha in matriz:
    print(linha)

media_linhas = []

for i in range(5):
    soma = 0
    for k in range(5):
        soma += matriz[i][k]
    media = soma / 5
    media_linhas.append(media)

print('\nA media de cada linha de sua matriz é:')
print(media_linhas)
```

```python
# 13. Solicite uma matriz 3×3 preenchida pelo usuário.
# Verifique (estrutura condicional) se ela é uma matriz identidade.

matriz = []

print('Quais são os valores da sua matriz 3x3?')
for i in range(3):
    linha = []
    for k in range(3):
        valores = float(input(f"ELEMENTO - [{i}][{k}]: "))
        linha.append(valores)
    matriz.append(linha)
print(f"\nA sua matriz é: ")

matriz_identidade = True

for linha in matriz:
    print(linha)

for i in range(3):
    for k in range(3):
        if i == k:
            if matriz[i][k] != 1:
                matriz_identidade = False
        else:
            if matriz[i][k] != 0:
                matriz_identidade = False
if matriz_identidade == True:
    print('\nA sua matriz é Identidade!')
else:
    print("\nA sua matriz não é identidade!")
```

```python
#16 - Crie uma matriz 4×4 com números fornecidos pelo usuário.
#Encontre o menor valor de cada coluna e armazene em uma lista separada.

matriz = []

print("Digite os valores da matriz 4x4")

for i in range(4):

    linha = []

    for k in range(4):
        valor = int(input(f"ELEMENTO [{i}][{k}]: "))
        linha.append(valor)

    matriz.append(linha)

menores = []

for k in range(4):

    menor = matriz[0][k]

    for i in range(4):

        if matriz[i][k] < menor:
            menor = matriz[i][k]

    menores.append(menor)

print("\nMatriz:")

for linha in matriz:
    print(linha)

print("\nMenores valores de cada coluna:")
print(menores)
```

```python
#17 - Dada uma matriz 3×3 com valores reais, calcule o determinante (sem usar bibliotecas externas).
#Use apenas listas e estruturas de repetição.

matriz = []

print("Digite os valores da matriz 3x3")

for i in range(3):

    linha = []

    for k in range(3):
        valor = float(input(f"ELEMENTO [{i}][{k}]: "))
        linha.append(valor)

    matriz.append(linha)

print("\nMatriz:")

for linha in matriz:
    print(linha)

determinante = (
    matriz[0][0] * matriz[1][1] * matriz[2][2]
    + matriz[0][1] * matriz[1][2] * matriz[2][0]
    + matriz[0][2] * matriz[1][0] * matriz[2][1]
    - matriz[0][2] * matriz[1][1] * matriz[2][0]
    - matriz[0][0] * matriz[1][2] * matriz[2][1]
    - matriz[0][1] * matriz[1][0] * matriz[2][2]
)

print("\nDeterminante:", determinante)
```

```python
#18 - Faça um programa que leia duas matrizes 3×3 e calcule a matriz soma e a matriz diferença, armazenando os resultados em novas listas.

matriz1 = []
matriz2 = []

print("Digite os valores da MATRIZ 1")

for i in range(3):

    linha = []

    for k in range(3):
        valor = int(input(f"MATRIZ 1 [{i}][{k}]: "))
        linha.append(valor)

    matriz1.append(linha)

print("\nDigite os valores da MATRIZ 2")

for i in range(3):

    linha = []

    for k in range(3):
        valor = int(input(f"MATRIZ 2 [{i}][{k}]: "))
        linha.append(valor)

    matriz2.append(linha)

soma = []
diferenca = []

for i in range(3):

    linha_soma = []
    linha_diferenca = []

    for k in range(3):

        linha_soma.append(matriz1[i][k] + matriz2[i][k])
        linha_diferenca.append(matriz1[i][k] - matriz2[i][k])

    soma.append(linha_soma)
    diferenca.append(linha_diferenca)

print("\nMatriz Soma:")

for linha in soma:
    print(linha)

print("\nMatriz Diferença:")

for linha in diferenca:
    print(linha)
```

```python
#Round III — Conjuntos (Sets)
#19 - Solicite dois conjuntos de 10 elementos fornecidos pelo usuário.
#Mostre a interseção, união e diferença (A − B e B − A).

conjuntoA = set()
conjuntoB = set()

for i in range(10):
    numero = int(input("Digite um número para o conjunto A: "))
    conjuntoA.add(numero)

for i in range(10):
    numero = int(input("Digite um número para o conjunto B: "))
    conjuntoB.add(numero)

print("União:")
print(conjuntoA | conjuntoB)

print("Interseção:")
print(conjuntoA & conjuntoB)

print("Diferença A - B:")
print(conjuntoA - conjuntoB)

print("Diferença B - A:")
print(conjuntoB - conjuntoA)
```

```python
#29 - Crie um dicionário onde as chaves são nomes de produtos e os valores são listas com preços praticados em três lojas diferentes.
#Encontre o produto com maior variação de preço.

produtos = {}

for i in range(3):
    nome = input("Nome do produto: ")

    precos = []

    for j in range(3):
        preco = float(input("Preço: "))
        precos.append(preco)

    produtos[nome] = precos

maior_variacao = 0
produto_maior = ""

for produto in produtos:
    variacao = max(produtos[produto]) - min(produtos[produto])

    if variacao > maior_variacao:
        maior_variacao = variacao
        produto_maior = produto

print("Produto com maior variação:", produto_maior)
print("Variação:", maior_variacao)
```

```python
#30 - Implemente um programa que leia uma frase e armazene em um dicionário o número de ocorrências de cada palavra (ignorando maiúsculas/minúsculas).
 #(Use estruturas de repetição.)

frase = input("Digite uma frase: ").lower()

frase += " "

dicionario = {}
palavra_atual = ""

for caractere in frase:

    if caractere != " ":
        palavra_atual += caractere

    elif palavra_atual != "":

        if palavra_atual in dicionario:
            dicionario[palavra_atual] += 1
        else:
            dicionario[palavra_atual] = 1

        palavra_atual = ""

print("\nOcorrências de cada palavra:")
for palavra, quantidade in dicionario.items():
    print(f"{palavra}: {quantidade}")
```
