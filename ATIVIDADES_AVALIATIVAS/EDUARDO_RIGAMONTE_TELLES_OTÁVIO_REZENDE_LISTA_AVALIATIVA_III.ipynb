```python
# 01. Crie uma lista com 15 números inteiros digitados pelo usuário.
# Em seguida, remova todos os números repetidos, mantendo a ordem original.

numeros = []
print("Digite 15 números inteiros:")

for i in range(15):
    num = int(input(f"Digite seu {i+1}° número: "))
    numeros.append(num)
    numeros.sort()

numeros_sem_repetir = []

for num in numeros:
    if num not in numeros_sem_repetir:
        numeros_sem_repetir.append(num)
        numeros_sem_repetir.sort()

print(f"\nLista Original - {numeros}")
print(f'\nLista sem números repetidos - {numeros_sem_repetir}')
```

```python
# 02. Escreva um programa que leia 10 nomes e 10 idades.
# Armazene os nomes em uma lista e as idades em outra.
# Em seguida, exiba o nome da pessoa mais velha.

nomes = []
idades = []
print("Digite 10 nomes:")

for i in range(10):
    nome = input(f"Digte o {i+1}° nome:  ").capitalize()
    idade = int(input(f"Digite a idade referente ao {i+1}° nome: "))
    nomes.append(nome)
    idades.append(idade)
    print("\n")
maior_idade = max(idades)
posicao_maior_idade = idades.index(maior_idade)
nome_da_pessoa = nomes[posicao_maior_idade]

print(f'A pessoa mais velha dentre as que estão na lista é {nome_da_pessoa} com a idade de {maior_idade} anos.')
```
