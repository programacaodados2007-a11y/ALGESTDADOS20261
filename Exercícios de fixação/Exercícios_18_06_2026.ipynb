```python
# 1 - Em um sistema de controle de acesso, é necessário verificar se um valor ultrapassa um limite definido.
# Elabore um programa em Python que solicite ao usuário a entrada de um número inteiro e determine se ele é maior que 10.
# Implemente a função verificar_maior_que_dez(numero), que receba um inteiro e retorne um valor booleano.
# O programa deve exibir uma mensagem descritiva ao usuário com base no resultado.

def verificar_maior_que_dez(numero):
    if numero > 10:
        return True
    else:
        return False

valor = int(input("Digte um número inteiro: "))
if verificar_maior_que_dez(valor):
    print("Maior que 10")
else:
    print("igual ou menor que 10")
```

```python
# 2 - Em sistemas de validação de dados numéricos, identificar se um valor é par ou ímpar é uma operação fundamental.
# Desenvolva um programa em Python que leia um número inteiro fornecido pelo usuário e informe sua paridade.
# Crie a função eh_par(numero) que retorne True se o número for divisível por 2 e False caso contrário.
# O programa deve exibir o resultado de forma clara.

def eh_par(numero):
    if numero % 2 == 0:
        return True
    else:
        return False

valor = int(input("Digte um valor inteiro: "))
if eh_par(valor):
    print(f"O número {valor} é par!")
else:
    print(f"O número {valor} não é par!")

# OR

print(eh_par(valor))
```

```python
# 3 - Em um sistema de cadastro, regras de validação de nomes são aplicadas automaticamente.
# Escreva um programa que leia um nome informado pelo usuário e verifique se ele contém a letra "a" (maiúscula ou minúscula).
# Implemente a função contem_letra_a(nome) que retorne um valor booleano conforme a condição.
# Exiba uma mensagem adequada ao usuário.

def contem_letra_a(nome):
    return "a" in nome.lower()

nome_do_usuario = input("Digte seu nome: ")

if contem_letra_a(nome_do_usuario):
    print("A sua palavra contém a letra a!")
else:
    print("A sua palavra não cotém a letra a!")

# or

# def contem_letra_a(nome):
#     if ("a" or "A") in nome:
#         return True
#     else:
#         return False

# palavra = input("Digite um nome: ")
# print(contem_letra_a(palavra))
```

```python
# 4 - Em sistemas de análise numérica, a divisibilidade é um critério frequentemente utilizado.
# Desenvolva um programa que leia um número inteiro e informe se ele é múltiplo de 5.
# Crie a função eh_multiplo_de_5(numero) que retorne um valor lógico com base na divisibilidade por 5.
# O programa deve exibir o resultado de forma interpretativa.

def eh_multiplo_de_5(numero):
    if numero % 5 == 0:
        return True
    else:
        return False

valor = int(input("Digte um valor inteiro: "))

if eh_multiplo_de_5(valor):
    print(f"O número {valor} é divisivel por 5!")
else:
    print(f"O número {valor} não é divisivel por 5!")
```
